```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data.csv')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print((df == 0).sum())
```

```python
df['price'] = df['price'].replace(0 , df['price'].median())
df['bedrooms'] = df['bedrooms'].replace(0 , df['bedrooms'].median())
df['bathrooms'] = df['bathrooms'].replace(0 , df['bathrooms'].median())
```

```python
# Price , Bedrooms , Bathrooms cannot be 0 , so i replace it with median()!
((df==0).sum())
```

```python
# x = input  , y = output , Goal = Predict House Price
df = df[df['price']< df['price'].quantile(0.99)]
```

```python
y = df['price']
```

```python
# MODEL 1 — SIMPLE LINEAR REGRESSION
x_lr = df[['sqft_living']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train_lr , x_test_lr , y_train_lr , y_test_lr = train_test_split(x_lr , y , test_size = 
                                                      0.2 , random_state = 42)
```

```python
print(x_train_lr.shape)
print(x_test_lr.shape)
print(y_train_lr.shape)
print(y_test_lr.shape)
```

```python
#Linear Regression
from  sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train_lr , y_train_lr)
pred_lr = lr.predict(x_test_lr)
pred_lr = np.maximum(pred_lr, 0)
print(pred_lr[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
mae_lr = mean_absolute_error(y_test_lr, pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test_lr, pred_lr))
r2_lr = r2_score(y_test_lr, pred_lr)

print("\n===== Linear Regression =====")
print('MAE:', mae_lr)
print('RMSE', rmse_lr)
print('R2_Score:', r2_lr)
```

```python
x_mlr = df.drop(
    ["price", "date", "street", "statezip", "country"],
    axis=1)
```

```python
x_mlr = pd.get_dummies(x_mlr, drop_first=True)
```

```python
x_train_mlr, x_test_mlr, y_train_mlr, y_test_mlr = train_test_split(
    x_mlr, y,
    test_size=0.2,
    random_state=42
)
```

```python
#Multiple Lineear Reression 
mlr = LinearRegression()

mlr.fit(x_train_mlr, y_train_mlr)

pred_mlr = mlr.predict(x_test_mlr)

pred_mlr = np.maximum(pred_mlr, 0)

mae_mlr = mean_absolute_error(y_test_mlr, pred_mlr)
rmse_mlr = np.sqrt(mean_squared_error(y_test_mlr, pred_mlr))
r2_mlr = r2_score(y_test_mlr, pred_mlr)

print("\n===== Multiple Linear Regression =====")

print("MAE:", mae_mlr)
print("RMSE:", rmse_mlr)
print("R2 Score:", r2_mlr)
```

Linear Regression was first applied using only one feature, sqft_living, to understand the basic relationship between house area and price. However, house price depends on multiple important factors such as bedrooms, bathrooms, location, floors, and condition. Therefore, Multiple Linear Regression was applied using multiple features, which significantly improved performance. The R² Score increased from 0.43 to 0.69, while MAE and RMSE decreased, proving that Multiple Linear Regression is the better model for House Price Prediction.

Answer:

Multiple Linear Regression performed best because house price depends on multiple independent variables, not just one. It achieved lower MAE and RMSE and a higher R² Score of 0.69 compared to Simple Linear Regression, making it more accurate and reliable.

```python
# Create Comparison Table.
comparison = pd.DataFrame({
    "Algorithm": ["Linear Regression", "Multiple Linear Regression"],
    
    "MAE": [mae_lr, mae_mlr],
    
    "RMSE": [rmse_lr, rmse_mlr],

    "R2 Score": [r2_lr, r2_mlr]
})

print(comparison)
print() 

print("Linear Regression: Uses only one feature (sqft_living), so prediction accuracy is lower with higher error values.")
print()

print("Multiple Linear Regression: Uses multiple important features, so prediction accuracy is better with lower error values and higher R2 Score.")
```

```python
#Showing Comparison using heatmap.
comparison = pd.DataFrame({
    "Algorithm": ["Linear Regression", "Multiple Linear Regression"],
    "MAE": [mae_lr, mae_mlr],
    "RMSE": [rmse_lr, rmse_mlr],
    "R2 Score": [r2_lr, r2_mlr]
})

comparison.set_index("Algorithm", inplace=True)

plt.figure(figsize=(8,5))
sns.heatmap(comparison, annot=True, fmt=".2f", cmap="YlGnBu")

plt.title("Model Performance Comparison Heatmap")
plt.show()
```
